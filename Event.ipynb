```python
# Four Basic Event Types
event_types = {
    '资产重组': 'Asset Restructuring' ,
    '对外担保': 'External Guarantee',
    '股份质押': 'Share Pledge',
    '资产收购': 'Asset Acquisition'
}
stock_gsrl_gsdt_em_df['Event_Type'] = stock_gsrl_gsdt_em_df['事件类型'].map(event_types)
set(stock_gsrl_gsdt_em_df['Event_Type'])
```

### Generate all trading days in 2023

```python
import datetime
import chinese_calendar

start_date = datetime.date(2023, 1, 1) # 2023-1-1
end_date = datetime.date(2023, 11, 24) # 2023-11-24
date_range = [start_date + datetime.timedelta(days=x) for x in range((end_date - start_date).days + 1)] # Generate Date range for 2023
trade_dates = [date for date in date_range if (chinese_calendar.is_workday(date) and date.weekday()>=0 and date.weekday()<=4)] # Filter Trading dates
print(trade_dates)
```

### Data Transform

```python
for date in trade_dates:
    datenumber = date.strftime("%Y%m%d")
    df = ak.stock_gsrl_gsdt_em(date=datenumber)
    print(datenumber)
    df['Event_Dim_ID'] = range(len(df))
    df['Event_Dim_ID'] += int(datenumber)*10000
    df['Stock_ID'] = df['代码'].astype(int)
    df['Stock_Name'] = df['简称']
    df['Event_Date'] = date
    df['Event_Type'] = df['事件类型'].map(event_types)
    df['Event_Description'] = df['具体事项']
    df['Current_Flag'] = 1
    df['Effective_Timestamp'] = date
    df['Expire_Timestamp'] = ''
    df = df[['Event_Dim_ID','Stock_ID','Stock_Name','Event_Date','Event_Type',
             'Event_Description','Current_Flag','Effective_Timestamp','Expire_Timestamp']]
    path ='./Dim_Event/'+datenumber+'.csv'
    df.to_csv(path,index=False, encoding='gbk')
    
```

### Update Current_Flag and Expire_Timstamp for SCD Type2

```python
total_df = total_df.sort_values(by='Event_Date')
total_df['Current_Flag'] = 1
total_df[:10]
```

```python
for index, row in total_df.iterrows():
    event_date = row["Event_Date"]
    stock_id = row["Stock_ID"]
    condition = (total_df["Event_Date"] < event_date) & (total_df["Stock_ID"] == stock_id) & (total_df["Expire_Timestamp"].isnull())
 
    # If their Expire_Timestamp is null, then update it with new the new date. And also upade Current_Flag
    total_df.loc[condition, "Expire_Timestamp"] = event_date
    total_df.loc[condition, "Current_Flag"] = 0
    print(index,stock_id,event_date,len(condition))
```
