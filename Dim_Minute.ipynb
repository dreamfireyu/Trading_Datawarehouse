```python
import os
import time
import pandas as pd
import akshare as ak
import warnings
from datetime import timedelta, datetime
warnings.filterwarnings("ignore")
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('expand_frame_repr', False)
pd.set_option('display.unicode.ambiguous_as_wide', True)
pd.set_option('display.unicode.east_asian_width', True)
```

```python
import datetime
import chinese_calendar

start_date = datetime.date(2023, 1, 1) # 2023-1-1
end_date = datetime.date(2023, 12, 31) # 2023-12-31
date_range = [start_date + datetime.timedelta(days=x) for x in range((end_date - start_date).days + 1)] # Generate Date range for 2023
trade_dates = [date for date in date_range if (chinese_calendar.is_workday(date) and date.weekday()>=0 and date.weekday()<=4)] # Filter Trading dates
print(trade_dates)
```

```python
trade_minutes = [] # store every minute in one day
for date in trade_dates: # iterate every trading day
    start_time1 = datetime.datetime(date.year, date.month, date.day, 9, 30) # 9:30
    end_time1 = datetime.datetime(date.year, date.month, date.day, 11, 30) # 11:30
    time_index = pd.date_range(start_time1, end_time1, freq='1min') # Generate time index
    start_time2 = datetime.datetime(date.year, date.month, date.day, 13, 1) # 13:01
    end_time2 = datetime.datetime(date.year, date.month, date.day, 15, 0) # 15:00
    time_index = time_index.append(pd.date_range(start_time2, end_time2, freq='1min')) 
    trade_minutes.extend(time_index) # Add time_index
print(trade_minutes)
```

```python
df = pd.DataFrame(trade_minutes, columns=['Date']) 
df['Minute_ID'] = df['Date'].dt.strftime('%Y%m%d%H%M')
df['Hour'] = df['Date'].dt.hour 
df['Minute'] = df['Date'].dt.minute 
df = df[['Minute_ID', 'Hour', 'Minute', 'Date']]
df
```

```python
df.to_csv('Dim_Minute.csv',index=False)
```
